```python
import numpy as np
from scipy.ndimage import rotate
from tensorflow.keras.datasets import mnist

#### Preprocessing ####
def generateTasks(X, y, task_type='permuted', num_tasks=10, seed=1):
   
    tasks = []    
    if task_type == 'permuted':
        np.random.seed(seed)
        for i in range(num_tasks):
            perm = np.random.permutation(784) # 784 pixels
            X_flat = X.reshape(len(X), -1)
            X_perm = X_flat[:, perm].reshape(-1, 28, 28) # 28x28 images
            tasks.append((X_perm, y))
            
    elif task_type == 'rotated':
        for i in range(num_tasks):
            angle = 10 * i
            X_rot = np.array([rotate(img, angle, reshape=False, order=1, mode='constant') for img in X])
            tasks.append((X_rot, y))
      
    return tasks

#### An example ####

# Load MNIST and normalize 
(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train = X_train / 255.0
X_test = X_test / 255.0

# Params, set for distinct task set - TODO maybe make a function that generates train and test tasks given parameters?
seed = 10 
num_tasks = 5  
task_type = 'permuted'

# Generate
train_perm_tasks = generateTasks(X_train, y_train, task_type = task_type, num_tasks=3, seed = seed)
test_perm_tasks = generateTasks(X_test, y_test, task_type = task_type, num_tasks=3, seed = seed)

# Access task 0 for example
X_perm_train_t0, y_perm_train_t0 = train_perm_tasks[0]
print(X_perm_train_t0.shape)
```
